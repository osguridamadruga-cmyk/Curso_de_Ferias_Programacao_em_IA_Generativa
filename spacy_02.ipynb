# Analise de texto 

```python
import spacy
import pandas as pd
from collections import Counter
```

```python
# Carrega o modelo em português
nlp  = spacy.load('pt_core_news_lg')

# Carrega o texto do arquivo .txt
with open('texto.txt', "r", encoding='utf-8') as arquivo:
    texto = arquivo.read()


# Processa o texto
doc = nlp(texto)
```

```python
# Lista onde as informações serão armazenadas 
dados = []

# Percorre todos os tokens
for token in doc:

    # Ignora pontuaçõa e espaços
    if token.is_punct or token.is_space:
        continue

    # Armazena as informações na lista
    dados.append({
         'Palavra' : token.text,
        "Lema" : token.lemma_,
        "Classe" : token.pos_,
        "Detalhe" : token.tag_
    })
```

```python
# Cria o DataFreme com as informaçõaes
df = pd.DataFrame(dados)
df
```

```python
df['Classe'].value_counts()
```

```python
frequencia = Counter(
    token.lemma_.lower()
    for token in doc
    if not token.is_punct
    and not token.is_space
    and not token.is_stop)

df_freq = pd.DataFrame(frequencia.items(), columns=['Palavra', 'Contagem'])
df_freq.sort_values('Contagem', ascending=False).head(20)
```
